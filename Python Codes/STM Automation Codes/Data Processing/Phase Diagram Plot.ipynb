```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
T_c = 7.5
H_c= 1.5
def phase(H_c,T):
    return H_c*(1-(T/T_c)**2)
```

```python
T = np.linspace(0, 7.5, 100)

for t in range(len(T)):
    sc= phase(H_c, T)
    plt.plot(T,sc)

plt.xlabel("Temperature (K)")
plt.ylabel("External Field (H)")
```
